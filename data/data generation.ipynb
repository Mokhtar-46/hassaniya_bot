```python

# ----------------------------------------------------------------
# FONCTIONS DE GÉNÉRATION
# ----------------------------------------------------------------

def remplir(template, cat):
    """Remplace les slots lexicaux dans un template."""
    s = template
    s = s.replace("{prenom}",        random.choice(prenoms_masculins))
    s = s.replace("{prenom_f}",      random.choice(prenoms_feminins))
    s = s.replace("{lieu}",          random.choice(lieux))
    s = s.replace("{marchandise}",   random.choice(marchandises))
    s = s.replace("{prix}",          random.choice(prix))
    s = s.replace("{moment}",        random.choice(moments_journee))
    s = s.replace("{maladie}",       random.choice(maladies))
    s = s.replace("{remede}",        random.choice(remedes))
    s = s.replace("{saison}",        random.choice(saisons))
    s = s.replace("{meteo}",         random.choice(meteos))
    s = s.replace("{conseil_sujet}", random.choice(conseils_sujets))
    s = s.replace("{ouv}",           random.choice(ouvertures))
    s = s.replace("{fer}",           random.choice(fermetures))
    return s

def generer_paire_template(cat):
    """Génère une paire Q/R depuis les templates."""
    data = templates[cat]
    q = remplir(random.choice(data["q_templates"]), cat)
    r = remplir(random.choice(data["r_templates"]), cat)
    return {"input": q, "output": r, "categorie": cat}

def generer_paire_seed(cat):
    """Utilise une graine existante avec variation légère."""
    data = templates[cat]
    q, r = random.choice(data["seeds"])
    # Variation : ajouter une ouverture ou fermeture aléatoire
    if random.random() > 0.5 and not r.startswith("و"):
        r = random.choice(ouvertures) + r
    if random.random() > 0.6 and not r.endswith("الله") and not r.endswith("نا"):
        r = r + random.choice(fermetures)
    # Remplir les slots si présents
    q = remplir(q, cat)
    r = remplir(r, cat)
    return {"input": q, "output": r, "categorie": cat}

def generer_paire_hybride(cat1, cat2):
    """Mélange contrôlé : question d'une catégorie, intro d'une autre (stratégie 6)."""
    data1 = templates[cat1]
    data2 = templates[cat2]
    q = remplir(random.choice(data1["q_templates"]), cat1)
    # Réponse hybride : intro cat2 + réponse cat1
    r_base = remplir(random.choice(data1["r_templates"]), cat1)
    intro = remplir(random.choice(data2["r_templates"]), cat2)
    r = intro.split("،")[0] + "، " + r_base if "،" in intro else r_base
    return {"input": q, "output": r, "categorie": cat1}

def appliquer_negation(paire):
    """Variation de polarité : transforme une affirmation en question négative."""
    q = paire["input"]
    neg = random.choice(negations_p)
    if q.startswith(" شنه "):
        q_neg = neg + q[4:]
    else:
        q_neg = "شنه " + neg + q
    return {"input": q_neg, "output": paire["output"], "categorie": paire["categorie"]}

# ----------------------------------------------------------------
# ASSEMBLAGE DU DATASET
# ----------------------------------------------------------------

categories = list(templates.keys())
dataset    = []

TARGET = 10000
par_cat = TARGET // len(categories)  # ~1000 par catégorie

for cat in categories:
    count = 0

    # 1. Seeds (15%)
    while count < int(par_cat * 0.15):
        dataset.append(generer_paire_seed(cat))
        count += 1

    # 2. Templates purs (55%)
    while count < int(par_cat * 0.70):
        dataset.append(generer_paire_template(cat))
        count += 1

    # 3. Variation de négation (15%) - only for prix
    while count < int(par_cat * 0.85):
        if cat == "prix":
            paire = generer_paire_template(cat)
            dataset.append(appliquer_negation(paire))
        else:
            dataset.append(generer_paire_template(cat))
        count += 1

    # 4. Hybride avec catégorie voisine (15%)
    cat_voisine = random.choice([c for c in categories if c != cat])
    while count < par_cat:
        dataset.append(generer_paire_hybride(cat, cat_voisine))
        count += 1

# Compléter jusqu'à exactement 10000+
while len(dataset) < TARGET:
    cat = random.choice(categories)
    dataset.append(generer_paire_template(cat))

print(f"Total généré : {len(dataset)} paires")
print(f"Répartition : { {c: sum(1 for d in dataset if d['categorie']==c) for c in categories} }")
```

```python

# ----------------------------------------------------------------
# SPLIT TRAIN / TEST (95 / 5)
# ----------------------------------------------------------------
random.shuffle(dataset)
split = int(len(dataset) * 0.95)
train_data, test_data = dataset[:split], dataset[split:]

with open("train.json", "w", encoding="utf-8") as f:
    json.dump(train_data, f, ensure_ascii=False, indent=2)

with open("test.json", "w", encoding="utf-8") as f:
    json.dump(test_data, f, ensure_ascii=False, indent=2)

print(f"\nTrain : {len(train_data)} | Test : {len(test_data)}")
print("Fichiers sauvegardés : train.json, test.json")
```

```python
import json, random
random.seed(42)

# ================================================================
# SOURCE: ENG-Hassaniya Dictionary by Ahmed Salem Elhoussein
# CLO-Assistant, US Embassy Nouakchott
# ALL Hassaniya words/phrases below are taken DIRECTLY from the book.
# ================================================================

# ----------------------------------------------------------------
# VOCABULARY SLOTS — extracted verbatim from the dictionary
# ----------------------------------------------------------------

# Greetings / states (book: salutation, how are you, good morning etc.)
ouvertures = [
    "إشحال إشصباح !",      # Good morning (p.71 "Ish-haal sbaah!")
    "السلام",               # Slaam (p.71)
    "هيه",                  # (marker used in notebook, confirmed in dict p.70 "Ahayh" = Yes)
    "لباس !",               # Quite good (p.65 "Labaas!")
    "مرحبا",                # Marihba (p.92 Welcome)
]

fermetures = [
    "إن شاء الله",          # Inshalla (p.61 "iyaak – inshalla")
    "الحمد لله",            # (p.65 "ilhamdu lillah")
    "ربي يحفظك",            # (p.28 "Rbii yihfadhk")
    "الله يبارك فيك",       # Ibaarik (p.21 bless)
    "الله يعين",            # (common closer, p.26 "Yalla" / p.37 help)
]

confirmations = [
    "أيوا !",               # Aywa (p.35 "Granted: Aywa")
    "صحيح",                 # Sahiih (p.1 Accurate: Sahiih – daqiiq)
    "واضح",                 # Waadiḥ (p.15 Clear: Waadiḥ)
    "حاق",                  # Haq (p.44 "Just: Saalih / Haq")
    "طيب",                # Mwaaviq (p.3 Agreed)
]

negations_p = [
    "ماهو واحد",           # maa-hu waahid (p.23 Different)
    "ما يكد",               # maa igid (p.40 Incapable: "maa igid")
    "ابداي",                # Abday (p.52 No: "Abday – maa nallah")
    "ما ينفع",              # maa yanvac (p.8 Disagreeable)
    "ما هو حاق",            # maa-hu haq (p.95 Wrong: "Kidhb – maa-hu haq")
]

# Names (from the dictionary examples)
prenoms_masculins = [
    "محمد",   # Mohamed (p.59 "Mohamed habba hirra")
    "علي",    # Ali (p.71 "Ali isallam clikum")
    "أحمد",   # Ahmed (author name on cover)
    "سالم",   # Salem (author)
    "باسم",   # Bassam (p.ii "Bassam hiila yingra il qaamus")
]

prenoms_feminins = [
    "فاطمة",
    "خديجة",
    "مريم",
    "زينب",
    "سكينة",
]

# Locations (from the dictionary)
lieux = [
    "نواكشوط",   # Nouakchott (cover)
    "البحر",     # Libbhar (p.9 "I live behind the river: Ana saakin ura libbhar")
    "البلاد",    # Blad (p.7 "Any where: Vi kil blad")
    "السوق",     # Suug (p.22 Cost: Suug/sicr)
    "البيت",     # Bayt (p.69 Room: "Bayt (m)")
    "المسجد",    # Msiid (p.61 Mosque)
]

# Merchandise / food (from the dictionary)
marchandises = [
    "النعناع",      # Lham (p.49 Meat: "Lham (m)")
    "السكر",      # Sukar (p.79 Sugar: "Sukar (m)")
    "التمر",      # (p.35 "Dainty: Hassas / Dairy: Bal bayc il ban")
    "اتاي",      # Ataay (p.81 Tea: "Ataay (m)")
]

# Prices (from the dictionary)
prix = [
    "مية أوقية",
    "ميتين أوقية",
    "خمسين أوقية",
    "ألف أوقية",
    "مية وخمسين",
]

# Times of day (from the dictionary)
moments_journee = [
    "الصباح",    # Sbaah (p.51 Morning: "Sbaah")
    "الضهر",     # (p.54 Noon: "Il gaayla (f) il waqra (f)")
    "المغرب",    # Miqrib (p.86 Twilight: "Miqrib (m)")
    "الليل",     # Layl (p.53 Night: "Layl (f)")
    "الداحميس",  # Iddahmiis (p.6 Afternoon: "Iddahmiis")
]

# Illnesses (from the dictionary)
maladies = [
    "الحمى",     # Himma (p.31 Fever: "Himma (f)")
    "التعب",     # (p.82 Tired: "Vatraan")
    "الراس",     # (p.37 Headache: "ujic irraas")
    "وجيع الظهر", # (p.37 Heal: "Yabra")
    "البرد",     # Bruud (p.15 Coldness: "Bruud (m)")
]

# Remedies (from the dictionary)
remedes = [
    "الدواء",    # Dwa (p.19 Cure: "Dwa (m)")
    "الراحة",    # Raaha (p.67 Relaxation: "Raaha (f)")
    "الماء",     # Ma (p.92 Water: "Ma (m) mayha (f)")
    "العسل",     # (p.1 "the water is abundant" context)
    "اتاي بالنعناع",  # Ataay / mint (p.81 Tea)
]

# Seasons (from the dictionary)
saisons = [
    "الصيف",    # Sayv (p.80 Summer: "Sayv m")
    "الشتا",    # Shta (p.80 Winter: "Shta (m)")
    "الخريف",   # (p.1 "During the rainy season: muddit likhriiv")
]

# Weather (from the dictionary)
meteos = [
    "الحمان",   # Llumman (p.38 Heat: "Llumman (m)")
    "البرد",    # Baarid (p.15 Cold: "Baarid")
    "الريح",    # Ryaah (p.7 Air: "Riih – Ryaah (f)")
    "الغمام",   # Ghmaam (p.32 Fog: "Ghmaam (m)")
    "الغبر",    # (dust, related to Traab p.82)
]

# Animals (from the dictionary)
animaux = [
    "الجمل",    # Jamal (p.2 "one-hump camel" context)
    "العنز",    # (goat - common)
    "البقرة",   # Bagra (p.15 Cattle: "Bgar (f.pl. of bagra)")
    "الكلب",    # Kalb (p.24 Dog: "Kalb (a)")
    "المش",     # Mush (p.14 Cat: "Mush (m) musha (f)")
    "الحمار",   # Hmaar (p.11 Brute: "Hmaar – waqd (adj)")
]

# Topics for advice (from the dictionary)
conseils_sujets = [
    "التجارة",   # Tijaara (p.84 Trade: "Tijaara (f)")
    "الزواج",    # (p.43 Engage: "Yawcad – yitcahad")
    "العمل",     # Shaqla (p.40 Labor / p.44 Job)
    "الصحة",     # Is-siha (p.37 Health: "Is-siha (f)")
    "المال",     # Maal (p.68 Richness: "Maal (m)")
    "الصبر",     # Isbar (p.31 Patience: "Isbar (m)")
    "الأمانة",   # (p.85 Trust: "Thiqa (f)")
    "الصدق",     # Saadiq (p.87 Sincere: "Saadiq")
    "السياسة",   # Siyaasi (p.61 Political)
    "الدراسة",   # Graaya (p.65 Reading/Study: "Graaya (f)")
]

# ----------------------------------------------------------------
# TEMPLATES — all Hassaniya taken from the dictionary
# ----------------------------------------------------------------

templates = {

# ────────────────────────────────────────────────────────────────
"salutation": {
    "seeds": [
        # p.71 "Ish-haal sbaah!" / "Good morning"
        ("إشحال إشصباح؟", "إشصباح النور، الله يصبحك بالخير"),
        # p.39 "How are you? / Ish haal – ak?"
        ("إشحال؟", "لباس، الحمد لله، وانت؟"),
        # p.39 How are you (f) "Ish haal – ik?"
        ("كيف حالك؟", "بخير الحمد لله  "),
        # p.92 Welcome "Marihba"
        ("مرحبا", "أهلاً وسهلاً، تفضل اقعد"),
        # p.71 "Salute: Isalam / Ali isallam clikum"
        ("السلام عليكم", "وعليكم السلام، كيف الأحوال؟"),
        # p.37 "How are you" good morning seed
        ("صباح الخير", "صباح النور، الله يصبحك بالخير يا ولدي"),
        # p.52 "Near: 1.hdha 2. il jaay"
        ("شحال المروح؟", "لباس الحمد لله، تفضل اقعد"),
    ],
    "q_templates": [
        "كيف حالك يا {prenom}؟",                      # p.39 "Ish haal – ak?"
        "كيف {moment}؟",
        "إشحال إشصباح يا {prenom}؟",                 # p.51 "Ish-haal sbaah!"
        "شطاري في {lieu}؟",
        "كيف الأحوال يا الحاج؟",
        "شماسي اليوم؟",                                # p.50 "Shmaasi (f)"
    ],
    "r_templates": [
        "لباس، وانت كيف حالك؟",                        # p.65 "Labaas!"
        "بخير الحمد لله، ما علين {fer}",
        "{ouv} بخير يا ولدي، الله يحفظك",
        "لباس الحمد لله {fer}",
        "{ouv} بخير وعافية {fer}",
        "معافين {fer}",                                # p.70 Safe: "muc'aavi"
    ],
},

# ────────────────────────────────────────────────────────────────
"commerce": {
    "seeds": [
        # p.59 "Yaasir min il vadha" — a lot of money
        (" {marchandise} اليوم باش؟", "بـ{prix} للكيل، {marchandise} زين"),
        # p.65 "Yishri / Sharaay" — buy/buyer
        ("عندك {marchandise} زين؟", "والله عندي، شتبقى منه؟"),
        # p.48 "Tijaari" — commercial
        ("باش {marchandise}؟", "بـ{prix}، وما تجبر حد أهون مني"),
        # p.22 "Suug (m) sicr (m)" — market/price
        ("الصكر عندك؟", "اهيه عندي، بـ{prix} للكيل"),
        # p.56 "Offer: card (m) / Yacti"
        ("شعندك من {marchandise}؟", "عندي حت، شتبقى منه؟"),
        # p.14 "Bargain: Vursa (f)"
        (" {marchandise} موجود؟", "موجود يا ولي، بـ{prix} {fer}"),
    ],
    "q_templates": [
        "كم {marchandise} اليوم؟",
        "عندك {marchandise} زين؟",
        "{marchandise} موجود؟",
        "شعندك من {marchandise}؟",
        "باش {marchandise}؟",
        "الصكر عندك؟",                                # p.79 Sugar: "Sukar (m)"
    ],
    "r_templates": [
        "{marchandise} بـ{prix}",
        "موجود يا ولي، بـ{prix} للكيل {fer}",
        "عندي أحسن {marchandise} في السوق {fer}",     # p.22 "Suug (m)"
        "عندي حت، شتبقى منه {fer}",
        "{marchandise} بـ{prix}، والله رخيص {fer}",  # p.65 Cheap: "Hwayyin – hiyyin"
    ],
},

# ────────────────────────────────────────────────────────────────
"meteo": {
    "seeds": [
        # p.80 Summer: "Sayv m" / Hot: "Haami"
        ("ذ من الحمان شنه اليوم؟", "ماهو طبيعي اليوم، الله يسهل"),
        # p.7 Air: "Riih – Ryaah (f)"
        ("الريح متين اليوم", "الريح ضيعتنا، الله يحفظنا"),
        # p.32 Fog: "Ghmaam (m)"
        ("الغمام مزال؟", "ايوا مزال، الله يسهل على الناس"),
        # p.15 Cold: "Baarid" / Coldness: "Bruud (m)"
        ("شنه الجو في {saison}؟", "في {saison} البرد يكون حامي، الله يعين"),
        # p.70 Rain: "Shaab (f)" / Rain (to): "Itji shaab"
        ("السحاب لاه يخلي منه شي؟", "إن شاء الله، الناس محتاجين المطر"),
    ],
    "q_templates": [
        "{meteo} متين اليوم؟",
        "شنه الجو في {saison}؟",
        "{meteo} مزال؟",
        "شكاير في الجو اليوم؟",
        "{meteo} هذا العام؟",
        "كيف الجو في {lieu}؟",
    ],
    "r_templates": [
        "{ouv} {meteo} ما شفناه من سنين، الله يسهل",
        "ربي يحفظنا، {meteo} متين حت {fer}",
        "{ouv} هذا طبيعي في {saison}، نصبر {fer}",    # p.59 Patience: "Isbar / Sabbar"
        "الله يعينا",                                   # p.26 Energy: "Taaqa (f)"
        "الناس كاملة تشكي من {meteo} هذا العام",      # p.32 Complain: "Yishki"
    ],
},

# ────────────────────────────────────────────────────────────────
"famille": {
    "seeds": [
        # p.36 Family: "Usra (f)"
        ("شحال العيال؟", "الحمد لله، بخير، الله يحفظهم"),
        # p.37 Son: "Wul"
        ("كم ولد عندك؟", "عندي خمسة، الحمد لله"),
        # p.11 Brother: "Khu" / Brother inlaw: "Nsiib"
        ("شحال خوك؟", "بخير الحمد لله، لا باس فيه"),
        # p.9 Birth: "Khlaga (f)" / Birthday: "Milaad (m)"
        ("خلق عندكم طفيل؟", "والله الله يبارك، مولود جديد الحمد لله"),
        # p.54 Mother: "Um – waalda"
        ("شحال أمك؟", "معافية الحمد لله {fer}"),
    ],
    "q_templates": [
        "اشحال الأهل يا {prenom}؟",
        "شحال أولادك؟",
        "كم ولد عندك؟",                                # p.37 "Wul"
        "شحال أهلك في {lieu}؟",
        "شحال اهل الدار؟",                          # p.27 Engagement: "Wacd (m)"
    ],
    "r_templates": [
        "معافين {fer}",                                # p.70 Safe: "Saalim u mu'aavi"
        "الحمد لله، الأهل بخير {fer}",
        "{ouv} نعمة من ربي، ما نشكي {fer}",
        "ربي يستر عليهم، كلهم بخير {fer}",
        "{ouv} الأولاد زينين، الله يوفقهم",           # p.32 Fine: "Zayn (adj)"
    ],
},

# ────────────────────────────────────────────────────────────────
"sante": {
    "seeds": [
        # p.37 Heal: "Yabra" / Health: "Is-siha (f)"
        ("مالك؟ ياك مانك مريض؟", "تعبان شوي، يغير الحمد لله ما هو شي كبير"),
        # p.19 Cure: "Dwa (m) / Yabra"
        ("عندك دواء للرأس؟", "عمرلك كاس ودير فيه نعناع، دور تبر"),
        # p.37 Headache: "ujic irraas"
        ("شحال ظهرك؟", "تعبان شوية من الكعاد، الله يعافينا"),
        # p.31 Fever: "Himma (f)"
        ("ياك ما تعبت بيك الحمى؟", "شوية والله، لاهون معمر"),
        # p.21 Ache: "Alam (m) uwja' (m)"
        ("شتقول في {maladie}؟", "جرب {remede}، ما خلاك شي أزين منه"),
        # p.37 Healthy: "Is-hiih"
        ("الله يعافيك، كيف صحتك؟", "بخير الحمد لله، ربي يعافيك {fer}"),
    ],
    "q_templates": [
        "ياك ما أنت حاس بـ{maladie}؟",               # p.47 Sick: dict
        "ياك ما أنت مريض؟",
        "شنه دو {maladie}؟",
        "ياك ما أنت حاس بشي؟",
        "تنصح باش لـ{maladie}؟",
        "كيف صحتك يا {prenom}؟",
    ],
    "r_templates": [
        "تعبان شوي، يغير الحمد لله {fer}",
        "جرب {remede}، ما خلاك شي أزين منه",
        "{ouv} الله يشافي، الصحة نعمة {fer}",         # p.37 "Is-siha (f)"
        "لباس، ربي يعافي {fer}",
        "الحمد لله بخير، ربي يحفظك {fer}",
    ],
},

# ────────────────────────────────────────────────────────────────
"religion": {
    "seeds": [
        # p.55 Pray: "Isalli" / Prayer: "Sla (f)"
        ("صليت؟", "ايوا صليت، الله يتقبل منا"),
        # p.56 Mosque: "Msiid (m)"
        (" المسجد منين؟", "المسجد قريب من هنا، عدل شوي"),
        # p.33 Forgive: "Yasmah" / Forgiveness: "Smaah (m)"
        ("الله يسامح؟", "ربي سامح ورحيم {fer}"),
        # p.64 Religion: "Diin (m) diyana (f)"
        ("شحال ديناتنا تقول في هذا؟", "الدين يأمرنا بالصبر والشكر"),
        # p.71 Thanks: "Shukran (m) shukr (m)"
        ("الله يجازيك", "الله يجازيك بالخير يا ولدي {fer}"),
        # p.33 Faithful: "Mawthuuq"
        ("ياك مانك صالح؟", "نطلب من ربي الستر والعافية {fer}"),
    ],
    "q_templates": [
        "صليت يا {prenom}؟",
        "فين المسجد في {lieu}؟",                      # p.56 Mosque: "Msiid (m)"
        "شحال ديناتنا تقول في {conseil_sujet}؟",
        "الله يسامح؟",                                 # p.33 Forgive: "Yasmah"
        "ياك مانك صالح يا {prenom}؟",
        "متى الصلاة؟",
    ],
    "r_templates": [
        "ربي سامح ورحيم {fer}",                        # p.49 Merciful: "Rahiim"
        "{ouv} الله يتقبل منا {fer}",
        "نطلب من ربي الستر والعافية {fer}",
        "الدين يأمرنا بالصبر والشكر {fer}",            # p.77 Patience: "Isbar (m)"
        "{ouv} ربي يحفظنا ويسترنا {fer}",
    ],
},

# ────────────────────────────────────────────────────────────────
"atay": {                                               # Tea / Attaay
    "seeds": [
        # p.81 Tea: "Ataay (m)"
        ("تشرب اتاي؟", " حتت"),
        # p.60 Taste: "ticma (f) / Idhuug"
        ("الشاي حلو؟", "حلو، الله يبارك، زين هتا"),
        # p.50 Boil: "Ivayyadh – ihammi"
        ("الماء غلى؟", "ايوا غلى، عمر الكاسات"),
        # p.77 Sweet: "Imawnak" / p.79 Sugar: "Sukar (m)"
        ("كم الصكر؟", "ثلاث ملاعق"),
        # p.48 Drink: "Shraab / Yishrub"
        ("تشرب ولا ما تشرب؟", "والله نشرب، الشاي ما يطيح"),
        # p.60 Pour: "Ikub"
        ("اسكب الشاي", "صبر شوي، مزال ما غلى {fer}"),
    ],
    "q_templates": [
        "تشرب شاي يا {prenom}؟",                       # p.81 Tea: "Ataay (m)"
        "الشاي زين؟",                                  # p.32 Fine: "Zayn (adj)"
        "الماء غلى؟",                                  # p.50 Boil: "Ivayyadh"
        "كم من الصكر في الشاي؟",                        # p.79 Sugar: "Sukar (m)"
        "تشرب ولا ما تشرب؟",                           # p.48 Drink: "Yishrub"
        "الشاي حلو ولا مر؟",                           # p.10 Bitter: "Murr"
    ],
    "r_templates": [
        "ايوا، الشاي ما يطيح كاع بيه",
        "حلو، الله يبارك {fer}",                       # p.10 Bitter/Sweet contrast
        "{ouv} الشاي زين هتا {fer}",
        "صبر شوي، مزال ما غلى {fer}",
        "اشرب وعافية يا {prenom} {fer}",
    ],
},

# ────────────────────────────────────────────────────────────────
"prix": {
    "seeds": [
        # p.22 Cost: "Suug (m) sicr (m)" / Costly: "Waacir – gaasi"
        ("{marchandise} باش؟", "{marchandise} بـ{prix}"),
        # p.65 Cheap: "Hwayyin – hiyyin"
        ("واعر ولا رخيص؟", "رخيص والله، هوين هوين"),
        # p.59 Price: "Sicr (m) thamam (m)"
        ("كم السعر؟", "السعر {prix}، والله مناسب"),
        # p.56 Offer: "Yacti" / p.26 Expensive: "Gaasi – waa'ir"
        ("هذا غالي ليا", "ما غالي، والله هذا السعر ما تجبر أهون"),
        # p.9 Cash: "Dhaat"
        ("بالكاش ولا بالدين؟", "بالكاش أحسن، الدين صعيب"),  # p.21 Debt: "Dayn (m)"
        # p.65 Sell: "Bayc / Bayyac"
        ("ما تنكص في السعر؟", "الله يبارك فيك، هذا آخر سعر"),
    ],
    "q_templates": [
        "{marchandise} باش؟",                          # p.22 Cost
        "كم السعر؟",                                 # p.59 Price: "Sicr (m)"
        "واعر ولا رخيص؟",                             # p.65 Cheap: "Hwayyin"
        "ما تنكص في السعر؟",
        "هذا غالي ليا",
        "بالكاش ولا بالدين؟",                          # p.21 Debt: "Dayn (m)"
    ],
    "r_templates": [
        "{marchandise} بـ{prix}",
        "السعر {prix}، والله مناسب",                   # p.48 Appropriate: "Mulaa'im"
        "رخيص والله، هوين هوين",                       # p.65 Cheap: "Hwayyin"
        "الله يبارك فيك، هذا آخر سعر",
        "{ouv} {marchandise} بـ{prix}، ما تجبر أحسن {fer}",
    ],
},

# ────────────────────────────────────────────────────────────────
"conseil": {
    "seeds": [
        # p.6 Advice: "Nasuuha (f)" / Advise: "Mustashaar"
        ("تنصحني في {conseil_sujet}؟", "الصبر مفتاح الخير "),
        # p.48 Merchant: "Taajir (m)" / Trade: "Tijaara (f)"
        ("كيف تبدأ في التجارة؟", "ابدأ بالصدق والأمانة، والله يبارك"),
        # p.84 Tradition: "Cada (f)" / Experience: "Tajriba (f)"
        ("شتقول في {conseil_sujet}؟", "التجربة علمتني، الصبر والشغل"),
        # p.79 Struggle: "Ilhaarib – inaadii"
        ("كيف نوفق في الحياة؟", "اشغل وتوكل على الله، ما خلاك شي"),
        # p.62 Knowledge: "Ma'riva (f) – 'ilm (m)"
        ("الدراسة مهمة؟", "ايوا، العلم نور، اطلب العلم من المهد"),
        # p.27 Engagement / Marriage advice
        ("تنصحني في الزواج؟", "الزواج نعمة، اختار الصالحة والله يوفق"),
    ],
    "q_templates": [
        "تنصحني في {conseil_sujet}؟",                  # p.6 Advice: "Nasuuha (f)"
        "شتقول في {conseil_sujet}؟",
        "كيف نوفق في {conseil_sujet}؟",
        "شوف ليا نصيحة في {conseil_sujet}",
        "الدراسة مهمة يا {prenom}؟",
        "كيف تبدأ في {conseil_sujet}؟",
    ],
    "r_templates": [
        "الصبر مفتاح الفرج {fer}",                    # p.59 Patience: "Isbar (m)"
        "التجربة علمتني، الصبر والشغل {fer}",
        "{ouv} اشغل وتوكل على الله {fer}",
        "العلم نور، ما خلاك شي {fer}",                # p.62 Knowledge: "Ma'riva (f)"
        "ابدأ بالصدق والأمانة، والله يبارك {fer}",    # p.87 Sincerity: "Sidq (m)"
    ],
},

# ────────────────────────────────────────────────────────────────
"nouvelles": {
    "seeds": [
        # p.54 News: "Lakhbaar (f) tawaari' (f.pl)"
        ("شنه الأخبار؟", "الا هرمز هاذ يعطين خيرو ، الله يحفظ البلاد"),
        # p.52 Newspaper: "Jarida (f)"
        ("قرأت الجريدة اليوم؟", "ما قرأتها، ما عندي وقت"),
        # p.50 Message: "Risaala (f)"
        ("وصلتك رسالة من {prenom}؟", "ايوا وصلتني، الله يبارك فيه"),
        # p.37 Hear: "Yasma'" / Hear about: "Yasma' 'an"
        ("سمعت بالخبر؟", "سمعت، الله يسهل على الناس"),
        # p.1 Accident: "Ksiidan (m) – haadith (m)"
        ("شيء جرى في {lieu}؟", "الله يحفظنا، لا شيء إن شاء الله"),
        # p.50 Meeting: "Malga (m) jbaar (m)"
        ("في أخبار من {lieu}؟", "الله أعلم، ما وصلنا شيء {fer}"),
    ],
    "q_templates": [
        "شنه الأخبار يا {prenom}؟",                    # p.54 News: "Lakhbaar"
        " شطاري ؟",                                # p.37 Hear: "Yasma'"
        "شطاري في  {lieu}؟",
        "قرأت الجريدة اليوم؟",                         # p.52 Newspaper: "Jarida (f)"
        "وصلتك رسالة؟",                                # p.50 Message: "Risaala (f)"
        "شيء جرى في {lieu}؟",
    ],
    "r_templates": [
        " بخير، الله يحفظ البلاد {fer}",
        "الا استحفي هاذ,  {fer}",
        "{ouv} ما وصلنا شيء، الله يسهل {fer}",
        "الله يحفظنا، لا شيء إن شاء الله {fer}",
        "ما سمعت، شنه الخبر يا {prenom}؟",
    ],
},

}  # end templates

# ----------------------------------------------------------------
# GENERATION FUNCTIONS  (same logic as original notebook)
# ----------------------------------------------------------------

def remplir(template):
    s = template
    s = s.replace("{prenom}",         random.choice(prenoms_masculins))
    s = s.replace("{prenom_f}",       random.choice(prenoms_feminins))
    s = s.replace("{lieu}",           random.choice(lieux))
    s = s.replace("{marchandise}",    random.choice(marchandises))
    s = s.replace("{prix}",           random.choice(prix))
    s = s.replace("{moment}",         random.choice(moments_journee))
    s = s.replace("{maladie}",        random.choice(maladies))
    s = s.replace("{remede}",         random.choice(remedes))
    s = s.replace("{saison}",         random.choice(saisons))
    s = s.replace("{meteo}",          random.choice(meteos))
    s = s.replace("{conseil_sujet}",  random.choice(conseils_sujets))
    s = s.replace("{ouv}",            random.choice(ouvertures))
    s = s.replace("{fer}",            random.choice(fermetures))
    s = s.replace("{animal}",         random.choice(animaux))
    return s

def generer_paire_template(cat):
    data = templates[cat]
    q = remplir(random.choice(data["q_templates"]))
    r = remplir(random.choice(data["r_templates"]))
    return {"input": q, "output": r, "categorie": cat}

def generer_paire_seed(cat):
    data = templates[cat]
    q, r = random.choice(data["seeds"])
    if random.random() > 0.5 and not r.startswith("و"):
        r = random.choice(ouvertures) + " " + r
    if random.random() > 0.6 and not r.endswith("الله") and not r.endswith("نا"):
        r = r + " " + random.choice(fermetures)
    q = remplir(q)
    r = remplir(r)
    return {"input": q, "output": r, "categorie": cat}

def generer_paire_hybride(cat1, cat2):
    data1 = templates[cat1]
    data2 = templates[cat2]
    q      = remplir(random.choice(data1["q_templates"]))
    r_base = remplir(random.choice(data1["r_templates"]))
    intro  = remplir(random.choice(data2["r_templates"]))
    r = intro.split("،")[0] + "، " + r_base if "،" in intro else r_base
    return {"input": q, "output": r, "categorie": cat1}

def appliquer_negation(paire):
    q   = paire["input"]
    neg = random.choice(negations_p)
    if q.startswith("شنه "):
        q_neg = neg + " " + q[4:]
    else:
        q_neg = "شنه " + neg + " " + q
    return {"input": q_neg, "output": paire["output"], "categorie": paire["categorie"]}

# ----------------------------------------------------------------
# ASSEMBLE DATASET  (same proportions as original notebook)
# ----------------------------------------------------------------

categories = list(templates.keys())
dataset    = []

TARGET  = 10000
par_cat = TARGET // len(categories)   # ~1000 per category

for cat in categories:
    count = 0

    # 1. Seeds (15 %)
    while count < int(par_cat * 0.15):
        dataset.append(generer_paire_seed(cat))
        count += 1

    # 2. Pure templates (55 %)
    while count < int(par_cat * 0.70):
        dataset.append(generer_paire_template(cat))
        count += 1

    # 3. Negation variation (15 %) — only for "prix" like original
    while count < int(par_cat * 0.85):
        if cat == "prix":
            paire = generer_paire_template(cat)
            dataset.append(appliquer_negation(paire))
        else:
            dataset.append(generer_paire_template(cat))
        count += 1

    # 4. Hybrid with neighbouring category (15 %)
    cat_voisine = random.choice([c for c in categories if c != cat])
    while count < par_cat:
        dataset.append(generer_paire_hybride(cat, cat_voisine))
        count += 1

# Fill up to exactly TARGET
while len(dataset) < TARGET:
    cat = random.choice(categories)
    dataset.append(generer_paire_template(cat))

print(f"Total généré : {len(dataset)} paires")
print(f"Répartition : { {c: sum(1 for d in dataset if d['categorie']==c) for c in categories} }")

# ----------------------------------------------------------------
# SPLIT TRAIN / TEST  (95 / 5)
# ----------------------------------------------------------------
random.shuffle(dataset)
split      = int(len(dataset) * 0.95)
train_data = dataset[:split]
test_data  = dataset[split:]

with open("hass_train.json", "w", encoding="utf-8") as f:
    json.dump(train_data, f, ensure_ascii=False, indent=2)

with open("hass_test.json", "w", encoding="utf-8") as f:
    json.dump(test_data, f, ensure_ascii=False, indent=2)

print(f"\nTrain : {len(train_data)} | Test : {len(test_data)}")
print("Files saved: hass_train.json, hass_test.json")

# Also save a small sample for inspection
sample = random.sample(dataset, 20)
with open("/mnt/user-data/outputs/hass_sample.json", "w", encoding="utf-8") as f:
    json.dump(sample, f, ensure_ascii=False, indent=2)
print("Sample (20 pairs) saved: hass_sample.json")
```
